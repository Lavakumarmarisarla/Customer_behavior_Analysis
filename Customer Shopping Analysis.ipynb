```python
dataset["Review Rating"]=dataset.groupby("Category")["Review Rating"].transform(lambda x: x.fillna(x.median()))
```

```python
dataset.columns=dataset.columns.str.lower()
dataset.columns=dataset.columns.str.replace(" ","_")
```

```python
dataset=dataset.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
dataset.columns
```

```python
#Create a column age group.
labels=["Young_Adult","Adult","Middle_Aged","Senior"]
dataset["age_group"]=pd.qcut(dataset["age"],q=4,labels=labels)
```

```python
dataset[["age","age_group"]].head()
```

```python
#Create Column Purchase_Frequency_Days.
frequency_mapping={'Fortnightly':14,
                   'Weekly':7,
                   'Monthly':30,
                   'Quarterly':90,
                   'Bi-Weekly':14,
                   'Annually':365,
                   'Every 3 Months':90}
dataset["purchase_frequency_days"]=dataset["frequency_of_purchases"].map(frequency_mapping)
```

```python
import pandas as pd
from sqlalchemy import create_engine, text
from sqlalchemy.engine import URL

# ✅ Load your dataset
dataset = pd.read_csv("C:\\Users\\lavakumar marisarla\\Downloads\\customer_shopping_behavior.csv")

# ✅ Connect to MySQL
engine = create_engine(URL.create(
    drivername="mysql+pymysql",
    username="root",
    password="Lava@2004",
    host="localhost",
    port=3306,
    database="customer_behavior"
))

# ✅ WRITE
with engine.begin() as conn:
    dataset.to_sql(name="customer", con=conn, if_exists="replace", index=False)
    print("✅ Written successfully")

# ✅ READ
with engine.connect() as conn:
    df = pd.read_sql_query(text("SELECT * FROM customer LIMIT 5"), con=conn)

df.head()
```
